# Pemilihan 100 Audio Acak dari `potong audio`
Notebook ini mengambil **100 file audio secara acak** dari folder **`potong audio`** tanpa mengubah isi maupun nama file aslinya, lalu menyalinnya ke folder **`100_NON SAWMIL`**.

```
import os
import random
import shutil

# ==============================================================================
# KONFIGURASI PATH
# ==============================================================================
SOURCE_DIR = "potong audio"
TARGET_DIR = "100_NON SAWMIL"
NUM_SAMPLES = 100

os.makedirs(TARGET_DIR, exist_ok=True)

print(f"[OK] Folder Sumber : {os.path.abspath(SOURCE_DIR)}")
print(f"[OK] Folder Tujuan : {os.path.abspath(TARGET_DIR)}")
print(f"[OK] Jumlah Target : {NUM_SAMPLES} audio acak")
```

---
### Eksekusi Pemilihan Acak & Penyalinan ke `100_NON SAWMIL`

```
# 1. Ambil seluruh daftar audio dari folder sumber
all_audio = [f for f in os.listdir(SOURCE_DIR) if f.lower().endswith(('.mp3', '.wav', '.ogg'))]
print(f"[INFO] Ditemukan {len(all_audio)} file audio di folder '{SOURCE_DIR}'.")

# 2. Bersihkan folder tujuan terlebih dahulu agar bersih
for f in os.listdir(TARGET_DIR):
    fp = os.path.join(TARGET_DIR, f)
    if os.path.isfile(fp):
        os.remove(fp)

# 3. Pilih 100 audio secara acak (tanpa merubah nama maupun isi file)
random.seed(42)  # Menjaga konsistensi reproduktibilitas
selected_files = random.sample(all_audio, min(NUM_SAMPLES, len(all_audio)))

# 4. Salin ke folder 100_NON SAWMIL
for fname in selected_files:
    src_path = os.path.join(SOURCE_DIR, fname)
    dst_path = os.path.join(TARGET_DIR, fname)
    shutil.copy2(src_path, dst_path)

print("="*60)
print(f"SELESAI! Berhasil memilih dan menyalin {len(selected_files)} audio acak.")
print(f"Folder tujuan: '{TARGET_DIR}' sekarang berisi {len(os.listdir(TARGET_DIR))} file audio.")
print("="*60)
```

---
### Verifikasi Hasil File di `100_NON SAWMIL`

```
# Tampilkan contoh 10 file pertama yang terpilih di folder 100_NON SAWMIL
files_in_target = sorted(os.listdir(TARGET_DIR), key=lambda x: int(os.path.splitext(x)[0]) if os.path.splitext(x)[0].isdigit() else 9999)
print(f"Total file di '{TARGET_DIR}': {len(files_in_target)}")
print("\nContoh 10 file pertama:")
for idx, f in enumerate(files_in_target[:10], 1):
    print(f" {idx:2d}. {f}")
print("...")
```
